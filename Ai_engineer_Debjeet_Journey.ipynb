```python
import time, random

def timer(func):
    def wrapper(*args, **kwargs):          # *args, **kwargs = ANY function wrap ho sakta hai
        t0 = time.time()
        result = func(*args, **kwargs)
        print(f"[timer] took {time.time()-t0:.3f}s")
        return result
    return wrapper

def retry(times):                          # yeh decorator FACTORY hai (parameter le raha hai)
    def real_decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, times + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"attempt {attempt} failed: {e}")
            raise RuntimeError("LLM down after retries!")
        return wrapper
    return real_decorator

@timer
def slow_load():
    time.sleep(0.5)
    return "data loaded"

@retry(times=3)
def fake_llm_call():
    if random.random() < 0.6:
        raise ConnectionError("429: too many requests!")
    return "LLM: The capital of India is New Delhi."

print(slow_load())       # timer se time print hoga
print(fake_llm_call())   # kabhi fail-retry hoga, kabhi success
```

```python
class Ticket:
  def __init__(self,titel,severity):
    self.titel = titel
    self.severity = severity
  def describe(self):
    return "["+ self.severity +"] "+ self.titel

t= Ticket("LLM hallucinates", "HIGH")
print(t.describe())

#title: str, source: str, content: str
```

```python
from dataclasses import dataclass

@dataclass
class Document:
    title: str
    source: str = "web"
    content: str = ""

d1 = Document(title="jeet", content="ai engineering", source="web")
d2 = Document(title="jeet", content="ai engineering", source="web")
print(d1)          # readable?
print(d1 == d2)    # True?
```

```python
class Embedder:
  def embed(self,text):
    return f"vector({len(text)}dims)"

class SearchEngine:
  def __init__(self):
    self.embedder= Embedder()
  def search(self,query):
    vec= self.embedder.embed(query)
    return f"{vec} =>doc"

s1= SearchEngine()
print(s1.search("what is llm"))
```

```python
# GOOGLE ka embedder aaj kal sasta hai
class GoogleEmbedder:
    def embed(self, text): return f"google_vec({len(text)})"

# कल nayi AI company ka embedder aaye
se2 = SearchEngine()
se2.embedder = GoogleEmbedder()   # ← bas swap — SearchEngine ka baaki code untouched!

print(se2.search("ai"))
```
